```python
# =========================
# 사전 정리
# =========================
# 충돌 줄이기 위해 캐시/잔존 패키지 확인 권장. 필요시 주석 해제:
# !pip -q uninstall -y torch torchvision torchaudio fastai || true

# CUDA 버전 확인 후 적합한 휠 선택 권장(Colab 최근 이미지는 CUDA 12.x가 일반적).[web:20][web:6]

# =========================
# 설치(고정 순서 유지)
# =========================
# 1) 전처리/형태소/중복제거
!pip -q install pandas openpyxl emoji tqdm datasketch JPype1 konlpy  # 전처리/형태소/중복제거 [web:1]

# 2) PyTorch 도메인 세트: 호환 매트릭스에 맞춘 핀 고정
#    옵션 A: torch 2.8 계열 (권장)
!pip -q install torch==2.8.0 torchvision==0.23.0 torchaudio==2.8.0  # 호환 세트 [web:6][web:12][web:16]

# 3) BERTopic과 주변 의존성
!pip -q install --no-deps bertopic                           # 의존성은 수동 관리 [web:13]
!pip -q install -U umap-learn hdbscan sentence-transformers plotly # BERTopic 주변 필수 [web:13][web:19]

# 4) fastai는 torch<2.9 요구 → 위 조합과 호환
!pip -q install fastai==2.8.4                                # 요구조건 준수 [web:5]

# =========================
# 공통 임포트/안전장치
# =========================
import io, os, re, sys, pandas as pd
from tqdm import tqdm
import numpy as np
import torch

# 결정성/성능
np.random.seed(42)                     # 재현성 기본 시드 [web:1]
torch.manual_seed(42)                  # Torch 시드 [web:1]
torch.use_deterministic_algorithms(False)  # 일부 CuBLAS 이슈 회피 [web:6]
torch.set_grad_enabled(False)          # 추론 전용 [web:6]

# tqdm 설정
try:
    from tqdm.notebook import tqdm as tqdm_nb
    tqdm_nb.pandas()
    tqdm = tqdm_nb
except Exception:
    tqdm.pandas()                      # progress_apply 활성화 [web:6]

# 설치 검증: 버전 프린트
print("torch:", torch.__version__)
try:
    import torchvision, torchaudio, fastai
    print("torchvision:", torchvision.__version__)
    print("torchaudio:", torchaudio.__version__)
    print("fastai:", fastai.__version__)
except Exception as e:
    print("Optional domain package check error:", e)


# =========================
# A: 뉴스/소셜 전처리
# =========================
from google.colab import files
from datasketch import MinHash, MinHashLSH
from urllib.parse import urlparse
from konlpy.tag import Okt

okt = Okt()

print("xlsx 또는 csv 파일(뉴스/소셜 혼합 가능)을 업로드하세요. (뉴스 파일은 1개 업로드 전제)")
uploaded = files.upload()
if not uploaded:
    raise SystemExit("업로드된 파일이 없습니다.")

def is_news(name: str) -> bool:
    n = name.lower()
    return ("news" in n) or ("뉴스" in n)

def read_any(name: str, content: bytes) -> dict:
    lower = name.lower()
    bio = io.BytesIO(content)
    if lower.endswith((".xlsx",".xlsm",".xlsb",".xls")):
        return pd.read_excel(bio, sheet_name=None, engine="openpyxl")
    elif lower.endswith(".csv"):
        for enc in [None, "utf-8-sig", "cp949"]:
            bio.seek(0)
            try:
                return {"Sheet1": pd.read_csv(bio, encoding=enc) if enc else pd.read_csv(bio)}
            except Exception:
                continue
        raise ValueError(f"CSV 인코딩 인식 실패: {name}")
    else:
        raise ValueError(f"지원하지 않는 확장자: {name}")

# 정규화 패턴
RE_DATETIME = [
    r"\b\d{4}[.\-/](?:0?[1-9]|1[0-2])[.\-/](?:0?[1-9]|[12]\d|3[01])\b",
    r"\b(?:0?[1-9]|1[0-2])[.\-/](?:0?[1-9]|[12]\d|3[01])\b",
    r"\b\d{4}년\s*0?[1-9]월\s*0?[1-9]일\b|\b\d{4}년\s*(?:1[0-2]|0?[1-9])월\s*(?:3[01]|[12]?\d)일\b",
    r"\b(?:오전|오후)\s*\d{1,2}시(?:\s*\d{1,2}분)?(?:\s*\d{1,2}초)?\b",
    r"\b\d{1,2}:\d{2}(?::\d{2})?\b",
]
RE_NUMERIC = [
    r"\b\d{1,3}(?:,\d{3})+(?:\.\d+)?\b",
    r"\b\d+\.\d+\b",
    r"\b\d+\b",
    r"\b\d+[%％]\b",
    r"\b\d{2,3}-\d{3,4}-\d{4}\b",
]
RE_UNITS = [r"\b\d+\s*(원|만원|억|조|시|분|초|월|일|년|호)\b"]

try:
    import emoji
    HAS_EMOJI = True
except Exception:
    HAS_EMOJI = False

def strip_datetime_numeric(x: str) -> str:
    if not isinstance(x, str): return ""
    s = x
    if HAS_EMOJI:
        try:
            s = emoji.demojize(s, language="ko")
        except:
            pass
    for rx in RE_DATETIME: s = re.sub(rx, " ", s)
    for rx in RE_UNITS: s = re.sub(rx, " ", s)
    for rx in RE_NUMERIC: s = re.sub(rx, " ", s)
    s = re.sub(r'http[s]?://\S+|www\.\S+', " ", s)
    s = re.sub(r'\S+@\S+\.\S+', " ", s)
    s = re.sub(r'[@＠]\w+', " ", s)
    s = re.sub(r'#(\w+)', r'\1', s)
    s = re.sub(r'([ㅋㅎㅠㅜㅡ])\1{2,}', r'\1\1', s)
    s = re.sub(r'([!?])\1+', r'\1', s)
    s = re.sub(r"[^0-9A-Za-z가-힣\s\.\,\!\?\:\;\[\]\-\%％]", " ", s)
    s = re.sub(r"\s+", " ", s).strip()
    return s

DOMAIN_MAP = {
    "twitter.com": "Twitter/X", "x.com": "Twitter/X", "t.co": "Twitter/X",
    "facebook.com": "Facebook", "instagram.com": "Instagram", "threads.net": "Threads",
    "youtube.com": "YouTube", "youtu.be": "YouTube", "reddit.com": "Reddit",
    "dcinside.com": "DCInside", "theqoo.net": "더쿠", "ppomppu.co.kr": "뽐뿌",
    "clien.net": "클리앙", "pann.nate.com": "네이트판", "instiz.net": "인스티즈",
    "ruliweb.com": "루리웹", "fmkorea.com": "에펨코리아",
    "naver.com": "NAVER", "blog.naver.com": "NAVER 블로그", "cafe.naver.com": "NAVER 카페",
    "daum.net": "Daum", "tistory.com": "티스토리", "cafe.daum.net": "Daum 카페",
}
def extract_domain(u: str) -> str:
    if not isinstance(u, str) or not u.strip(): return ""
    s = "http://" + u if not re.match(r'^[a-zA-Z][a-zA-Z0-9+\-.]*://', u) else u
    try:
        netloc = urlparse(s).netloc.lower().replace("www.", "")
        return netloc.split(":")[0]
    except:
        return ""
def map_source(u: str) -> str:
    d_str = extract_domain(u)
    if not d_str: return "Unknown"
    if d_str in DOMAIN_MAP: return DOMAIN_MAP[d_str]
    parts = d_str.split(".")
    for i in range(len(parts)-2, -1, -1):
        cand = ".".join(parts[i:])
        if cand in DOMAIN_MAP: return DOMAIN_MAP[cand]
    return "Other"

def convert_date_14_to_YmdH(series: pd.Series) -> pd.Series:
    s = series.astype(str).str.extract(r'(\d{14})', expand=False)
    ts = pd.to_datetime(s, format="%Y%m%d%H%M%S", errors="coerce")
    return ts.dt.strftime("%Y.%m.%d.%H")

def is_ad(title, content):
    if not isinstance(title, str): title = ""
    if not isinstance(content, str): content = ""
    full = title + " " + content
    if any(kw in title for kw in ['[광고]','[홍보]','[추천]']): return True
    if any(kw in full for kw in ['변호사','법무법인','상담센터','법률사무소']) and any(kw in full for kw in ['상담','문의','비용','카카오톡','카톡','전화']):
        return True
    if any(kw in full for kw in ['강제추행/ 카메라촬영죄/ 아청법/','법률상담','카카오톡 채팅상담','pf.kakao.com','공구','공동구매','세트 구매','#공구오픈','#릴스','모집','공고','채용']):
        return True
    return False

SIMILARITY_THRESHOLD = 0.90
NUM_PERMUTATIONS = 128

def dedup_minhash_lsh(df, text_col):
    if df is None or df.empty or len(df) < 2: return df
    d = df.copy().reset_index(drop=True)
    texts = d[text_col].fillna("").astype(str).tolist()
    signatures = []
    for text in tqdm(texts, desc="MinHashing"):
        mh = MinHash(num_perm=NUM_PERMUTATIONS)
        for sh in set(text[i:i+3] for i in range(max(len(text)-2, 1))):
            mh.update(sh.encode('utf8'))
        signatures.append(mh)
    lsh = MinHashLSH(threshold=SIMILARITY_THRESHOLD, num_perm=NUM_PERMUTATIONS)
    for i, sig in enumerate(signatures):
        lsh.insert(i, sig)
    clusters, used = {}, set()
    for i, sig in enumerate(tqdm(signatures, desc="Clustering")):
        if i in used: continue
        q = lsh.query(sig)
        clusters[i] = q
        used.update(q)
    keep = sorted(clusters.keys())
    return d.loc[keep].reset_index(drop=True)

STOPWORDS_SOCIAL = set(["이번","것","수","등","더","관련","부분","문제","사용","정보","정도","때문","오늘","최근","기사","영상","사진","진짜","정말","그냥","이건","저건"])
NEWS_STOPWORDS = set(["은","는","이","가","을","를","과","와","도","만","로","으로","에","에서","하다","했다","한다","되다","됐다","된다","위해","통해","따라","대한","관련","기자","뉴스","사진","지난","올해","이번","최근","때문","정도","부분","문제"])
RE_NUM_TOKEN = re.compile(r"^\d+[a-zA-Z가-힣]*$")

def post_filter_tokens(tokens, stopset):
    out = []
    for t in tokens:
        if len(t) < 2: continue
        if RE_NUM_TOKEN.match(t): continue
        if t in stopset: continue
        out.append(t)
    return out

def extract_nouns_social(text):
    base = text if isinstance(text,str) else ""
    nouns = okt.nouns(base)
    nouns = post_filter_tokens(nouns, STOPWORDS_SOCIAL)
    return " ".join(nouns)

def extract_morphs_news(text):
    base = text if isinstance(text,str) else ""
    morphs = okt.morphs(base, stem=True)
    morphs = post_filter_tokens(morphs, NEWS_STOPWORDS)
    return " ".join(morphs)

def preprocess_social_df(df, title_col="제목", content_col="본문", date_col="날짜", url_col="url"):
    d = df.copy()
    if content_col in d.columns:
        d = d.dropna(subset=[content_col])
    if date_col in d.columns:
        d["date"] = convert_date_14_to_YmdH(d[date_col])
        d = d.drop(columns=[date_col], errors="ignore")
    if url_col in d.columns:
        d["source"] = d[url_col].astype(str).apply(map_source)
        d = d.drop(columns=[url_col], errors="ignore")
    d[f"{content_col}_clean"] = d[content_col].progress_apply(strip_datetime_numeric) if content_col in d.columns else ""
    d[f"{title_col}_clean"] = d[title_col].progress_apply(strip_datetime_numeric) if title_col in d.columns else ""
    if f"{content_col}_clean" in d.columns and len(d) > 1:
        d = dedup_minhash_lsh(d, f"{content_col}_clean")
    if title_col in d.columns and f"{content_col}_clean" in d.columns:
        flags = d.apply(lambda r: is_ad(r.get(f"{title_col}_clean",""), r.get(f"{content_col}_clean","")), axis=1)
        d = d[~flags].reset_index(drop=True)
    if f"{content_col}_clean" in d.columns and not d.empty:
        d["tok_len"] = d[f"{content_col}_clean"].apply(lambda x: len(str(x).split()))
        thr = int(d["tok_len"].quantile(0.95))
        d = d[d["tok_len"] <= thr].drop(columns=["tok_len"]).reset_index(drop=True)
    if f"{content_col}_clean" in d.columns and not d.empty:
        d["noun_text"] = d[f"{content_col}_clean"].progress_apply(extract_nouns_social)
    else:
        d["noun_text"] = ""
    return d

def preprocess_news_df(df, title_col="title", content_col="content", date_col="date", url_col="url"):
    d = df.copy()
    if content_col in d.columns:
        d = d.dropna(subset=[content_col])
    if date_col in d.columns:
        d["date_formatted"] = convert_date_14_to_YmdH(d[date_col])
        d = d.drop(columns=[date_col], errors="ignore")
    if url_col in d.columns:
        d = d.drop(columns=[url_col], errors="ignore")
    d["content_clean"] = d[content_col].progress_apply(strip_datetime_numeric) if content_col in d.columns else ""
    if not d.empty:
        d["morph_text"] = d["content_clean"].progress_apply(extract_morphs_news)
    else:
        d["morph_text"] = ""
    return d

social_parts, news_parts = [], []
def safe_var_name(pathname: str) -> str:
    base = os.path.splitext(os.path.basename(pathname))[0]
    var = re.sub(r"[^0-9A-Za-z가-힣_]+", "_", base)
    if re.match(r"^[0-9]", var):
        var = f"f_{var}"
    return var

raw_df_created = False
for fname, content in uploaded.items():
    print(f"\n=== 처리 시작: {fname} ===")
    try:
        sheets = read_any(fname, content)
    except Exception as e:
        print(f"로드 실패: {e}")
        continue

    if is_news(fname) and not raw_df_created:
        raw_concat_news = pd.concat(list(sheets.values()), ignore_index=True) if sheets else pd.DataFrame()
        varname = f"raw_{safe_var_name(fname)}_data"
        globals()[varname] = raw_concat_news.copy()
        print(f"[저장] {varname} (rows={len(raw_concat_news)})")
        raw_df_created = True
        for sheet_name, df in sheets.items():
            df_news = preprocess_news_df(df, title_col="title", content_col="content", date_col="date", url_col="url")
            df_news["_source_file"] = os.path.basename(fname); df_news["_sheet"] = sheet_name
            news_parts.append(df_news)
    else:
        for sheet_name, df in sheets.items():
            df_soc = preprocess_social_df(df, title_col="제목", content_col="본문", date_col="날짜", url_col="url")
            df_soc["_source_file"] = os.path.basename(fname); df_soc["_sheet"] = sheet_name
            social_parts.append(df_soc)

social_df_all = pd.concat(social_parts, ignore_index=True) if social_parts else None
news_df_all = pd.concat(news_parts, ignore_index=True) if news_parts else None

if raw_df_created: print("\n[완료] raw_<파일명>_data 변수가 생성되었습니다.")
else: print("\n[알림] 뉴스 파일이 없거나 생성에 실패했습니다.")
print(f"[완료] social_df_all: {None if social_df_all is None else social_df_all.shape}")
print(f"[완료] news_df_all: {None if news_df_all is None else news_df_all.shape}")

# =========================
# B: 토픽모델링 + 입법예고 + 시계열 (+ df_finaltest; label0 선집계)
# =========================
import io, re, torch
import pandas as pd
import numpy as np
from google.colab import files
from sentence_transformers import SentenceTransformer, util
from bertopic import BERTopic
from umap import UMAP

# raw_<파일명>_data 자동 감지
raw_vars = [k for k in globals().keys() if k.startswith("raw_") and k.endswith("_data")]
if not raw_vars:
    raise SystemExit("raw_<파일명>_data 변수를 찾지 못했습니다. 전처리 단계 확인 필요.")
if len(raw_vars) > 1:
    print("[경고] raw_<파일명>_data 변수가 2개 이상 감지됨. 첫 번째 사용:", raw_vars)
raw_var_name = raw_vars[0]
df_raw = globals()[raw_var_name].copy()
file_name = raw_var_name.removeprefix("raw_").removesuffix("_data")
print(f"[진단] 사용 DF 변수: {raw_var_name}, 파일명 키: {file_name}")

# 컬럼 매핑
def pick_text_column(df: pd.DataFrame) -> str:
    TEXT_CANDIDATES = ["content","본문","text","article","body"]
    lower_map = {c.lower(): c for c in df.columns}
    for cand in TEXT_CANDIDATES:
        if cand in lower_map: return lower_map[cand]
    str_cols = [c for c in df.columns if pd.api.types.is_string_dtype(df[c])]
    if not str_cols: raise ValueError("텍스트 컬럼을 찾을 수 없습니다.")
    lens = {c: df[c].astype(str).str.len().mean() for c in str_cols}
    return max(lens, key=lens.get)

col_map_candidates = {
    "date": ["raw_date","date","날짜","DATE"],
    "title": ["title","제목","TITLE"],
    "content": ["content","본문","text","article","body","CONTENT"]
}
def first_existing(df, names):
    for n in names:
        if n in df.columns: return n
    return None

date_col    = first_existing(df_raw, col_map_candidates["date"])
content_col = first_existing(df_raw, col_map_candidates["content"]) or pick_text_column(df_raw)

if "date_formatted" in df_raw.columns: df_raw = df_raw.drop(columns=["date_formatted"])
if date_col is not None and date_col != "date":
    df_raw = df_raw.rename(columns={date_col: "date"})
elif date_col is None:
    df_raw["date"] = None

# 파일명 기반 파라미터
law_param_map = [
    (["정보통신법","정보통신"], 15, 20, 5),
    (["개인정보보호법","개인정보"], 10, 25, 10),
    (["아동복지법","아동복지"], 5, 10, 5),
    (["자본시장법","자본시장"], 10, 15, 10),
]
def get_params_from_name(name: str):
    n = (name or "").lower()
    for keys, mts, nn, nc in law_param_map:
        if any(k.lower() in n for k in keys):
            return dict(min_topic_size=mts, n_neighbors=nn, n_components=nc)
    return dict(min_topic_size=10, n_neighbors=15, n_components=5)

params = get_params_from_name(file_name)
print(f"[매핑] min_topic_size={params['min_topic_size']}, n_neighbors={params['n_neighbors']}, n_components={params['n_components']}")

# 입력 생성
df_clean_input = df_raw.dropna(subset=[content_col]).copy()
if content_col != "content": df_clean_input["content"] = df_clean_input[content_col].astype(str)
docs = df_clean_input["content"].astype(str).tolist()
print(f"[진단] 텍스트 컬럼: content, 문서 수: {len(docs)}")

# 임베딩/BERTopic
embedding_model = SentenceTransformer("snunlp/KR-SBERT-V40K-klueNLI-augSTS")
embedding_model._first_module().eval()
umap_model = UMAP(n_neighbors=params["n_neighbors"], n_components=params["n_components"], min_dist=0.0, metric="cosine", random_state=42)
topic_model = BERTopic(embedding_model=embedding_model, umap_model=umap_model, min_topic_size=params["min_topic_size"], nr_topics="auto", verbose=True)

print("\n[학습] BERTopic 모델 학습 시작")
topics, probs = topic_model.fit_transform(docs)
print("[학습] 완료")

# 산출
df_cleaned = df_clean_input.copy()
df_cleaned["topic_id"] = topics
if probs is not None:
    try: df_cleaned["topic_prob"] = probs
    except Exception: pass

df_topics_only = df_cleaned[df_cleaned["topic_id"] != -1].copy()
topic_aggregated_df = (
    df_topics_only.groupby("topic_id", as_index=False)
    .agg(topic_text=("content", lambda x: " ".join(x.astype(str))))
)

doc_info_df = topic_model.get_document_info(docs)

# 입법예고 업로드
if "bills_df" not in globals() or bills_df is None or len(bills_df) == 0:
    print("\n입법예고 CSV를 업로드하세요.")
    uploaded_bills = files.upload()
    bills_file_name = next(iter(uploaded_bills))
    print(f"[업로드 완료] {bills_file_name}")
    try:
        bills_df = pd.read_csv(io.BytesIO(uploaded_bills[bills_file_name]))
    except UnicodeDecodeError:
        bills_df = pd.read_csv(io.BytesIO(uploaded_bills[bills_file_name]), encoding="cp949")

for col in ["법률안명","주요내용"]:
    if col not in bills_df.columns: bills_df[col] = ""
bills_df["법률안명"] = bills_df["법률안명"].fillna("")
bills_df["주요내용"] = bills_df["주요내용"].fillna("")
bills_df["bill_text"] = bills_df["법률안명"] + " " + bills_df["주요내용"]

print("\n[bills_df 컬럼 목록]")
print(list(bills_df.columns))
print("\n[샘플 5행]")
print(bills_df.head(5))

# 날짜 파싱: 자동+수동+float 특례
BILL_DATE_COL = "date"   # 필요시 정확한 컬럼명으로 변경
date_candidates = [
    "bill_date","date","발의일","공고일","입법예고일","입법예고기간시작일","입법예고기간종료일",
    "등록일","작성일","게시일","시행일","공포일","접수일","상정일","의결일","시작일","종료일",
    "Date","DATE","created_at","createdAt","created","published","published_at","publishedAt"
]
avail = [c for c in date_candidates if c in bills_df.columns]
if not avail:
    avail = [c for c in bills_df.columns if bills_df[c].dtype == "object"]

def parse_any_date(s):
    s = "" if pd.isna(s) else str(s).strip()
    if not s: return pd.NaT
    nums = re.sub(r"[^0-9]", "", s)
    for fmt, cond in [("%Y%m%d%H%M%S", len(nums)==14), ("%Y%m%d%H%M", len(nums)==12), ("%Y%m%d", len(nums)==8), ("%Y%m", len(nums)==6), ("%Y", len(nums)==4)]:
        if cond:
            dt = pd.to_datetime(nums, format=fmt, errors="coerce")
            if pd.notna(dt): return dt
    for pat, fmt in [
        (r"^\d{4}-\d{2}-\d{2}$", "%Y-%m-%d"),
        (r"^\d{4}\.\d{2}\.\d{2}$", "%Y.%m.%d"),
        (r"^\d{4}/\d{2}/\d{2}$", "%Y/%m/%d"),
        (r"^\d{4}-\d{2}-\d{2}\s+\d{2}:\d{2}(:\d{2})?$", "%Y-%m-%d %H:%M:%S"),
        (r"^\d{4}\.\d{2}\.\d{2}\s+\d{2}:\d{2}(:\d{2})?$", "%Y.%m.%d %H:%M:%S"),
        (r"^\d{4}년\s*\d{1,2}월\s*\d{1,2}일$", None),
    ]:
        if re.match(pat, s):
            if fmt is not None:
                dt = pd.to_datetime(s, format=fmt, errors="coerce")
                if pd.notna(dt): return dt
            else:
                nums2 = re.sub(r"[^0-9]", " ", s); parts = [p for p in nums2.split() if p.isdigit()]
                if len(parts) >= 3:
                    y, m, d = parts[0], parts[1], parts[2]
                    try:
                        return pd.to_datetime(f"{int(y):04d}-{int(m):02d}-{int(d):02d}", errors="coerce")
                    except:
                        pass
    return pd.to_datetime(s, errors="coerce", utc=False)

def to_ymd8(v):
    if pd.isna(v): return ""
    try: return f"{int(float(v)):08d}"
    except:
        s = re.sub(r"[^0-9]","", str(v))
        return s[:8] if len(s) >= 8 else ""

bill_date_series = None
if BILL_DATE_COL and BILL_DATE_COL in bills_df.columns:
    if BILL_DATE_COL == "date":
        tmp = bills_df[BILL_DATE_COL].apply(to_ymd8)
        bill_date_series = pd.to_datetime(tmp, format="%Y%m%d", errors="coerce")
        print("[수동+특례] 'date' 열(숫자형 YYYYMMDD)을 8자리로 변환하여 bill_date 생성")
    else:
        bill_date_series = bills_df[BILL_DATE_COL].apply(parse_any_date)
    print(f"[수동] bill_date source='{BILL_DATE_COL}'")
else:
    scores, parsed_map = [], {}
    for c in avail:
        if c == "date":
            p = bills_df[c].apply(lambda v: pd.to_datetime(to_ymd8(v), format="%Y%m%d", errors="coerce"))
        else:
            p = bills_df[c].apply(parse_any_date)
        parsed_map[c] = p
        ratio = p.notna().mean()
        span = (p.max() - p.min()).days if p.notna().sum() > 1 else 0
        score = ratio + (0.05 if span > 0 else 0.0)
        scores.append((c, score, ratio, span))
    if scores:
        scores.sort(key=lambda x: x[1], reverse=True)
        best_c, best_score, best_ratio, best_span = scores[0]
        if best_ratio > 0:
            bill_date_series = parsed_map[best_c]
            print(f"[자동] bill_date source='{best_c}', notna ratio={best_ratio:.3f}, span(days)={best_span}")
    if bill_date_series is None and avail:
        combined = bills_df[avail].astype(str).agg(" ".join, axis=1)
        p = combined.apply(parse_any_date)
        if p.notna().any():
            bill_date_series = p
            print(f"[백업] bill_date source='combined {len(avail)} cols', notna ratio={p.notna().mean():.3f}")

if bill_date_series is None:
    print("[warn] bill_date 파싱 실패: BILL_DATE_COL을 확인하거나 파일의 날짜 포맷을 점검하세요.")
else:
    bills_df["bill_date"] = bill_date_series
    mask_valid = bills_df["bill_date"].between(pd.Timestamp("1900-01-01"), pd.Timestamp("2100-12-31"))
    bills_df.loc[~mask_valid, "bill_date"] = pd.NaT
    print("[ok] bill_date notna ratio:", bills_df["bill_date"].notna().mean())
    print(bills_df[["bill_date"] + [c for c in avail[:3]]].head(8))

# 매칭(유사도는 CPU 계산)
print("\n[매칭] 토픽-법안 임베딩 계산 및 코사인 유사도")
topic_embeddings = embedding_model.encode(topic_aggregated_df["topic_text"].tolist(), convert_to_tensor=True, show_progress_bar=True)
bill_embeddings  = embedding_model.encode(bills_df["bill_text"].tolist(), convert_to_tensor=True, show_progress_bar=True)
cosine_scores = util.pytorch_cos_sim(topic_embeddings.detach().to("cpu"), bill_embeddings.detach().to("cpu"))

results = []
for i in range(len(topic_aggregated_df)):
    topic_scores = cosine_scores[i]
    best_score, best_idx = torch.max(topic_scores, dim=0)
    results.append({
        "topic_id": int(topic_aggregated_df.loc[i, "topic_id"]),
        "topic_text_preview": topic_aggregated_df.loc[i, "topic_text"][:150] + "...",
        "best_match_bill_name": bills_df.loc[int(best_idx.item()), "법률안명"],
        "best_match_bill_content": bills_df.loc[int(best_idx.item()), "bill_text"][:200] + "...",
        "similarity_score": float(best_score.item())
    })
results_df = pd.DataFrame(results)

THRESHOLD = 0.6
one_to_n_matches = []
for i in range(len(topic_aggregated_df)):
    topic_scores = cosine_scores[i]
    matched_bill_indices = (topic_scores >= THRESHOLD).nonzero(as_tuple=True)[0]
    if len(matched_bill_indices) > 1:
        matched_bill_names = bills_df.loc[matched_bill_indices.tolist(), "법률안명"].tolist()
        one_to_n_matches.append({"topic_id": int(topic_aggregated_df.loc[i, "topic_id"]), "matched_bill_count": len(matched_bill_names), "matched_bill_names": matched_bill_names})
one_to_n_df = pd.DataFrame(one_to_n_matches).sort_values(by="matched_bill_count", ascending=False) if one_to_n_matches else pd.DataFrame()

all_matched_pairs = (cosine_scores >= THRESHOLD).nonzero(as_tuple=False).tolist()
bill_to_topics_map, n_to_one_records = {}, []
for topic_idx, bill_idx in all_matched_pairs:
    bill_name = bills_df.loc[bill_idx, "법률안명"]; topic_id = int(topic_aggregated_df.loc[topic_idx, "topic_id"])
    bill_to_topics_map.setdefault(bill_name, []).append(topic_id)
for bill_name, topic_ids in bill_to_topics_map.items():
    uniq = sorted(set(topic_ids))
    if len(uniq) > 1:
        n_to_one_records.append({"bill_name": bill_name, "matched_topic_count": len(uniq), "matched_topic_ids": uniq})
n_to_one_df = pd.DataFrame(n_to_one_records).sort_values(by="matched_topic_count", ascending=False) if n_to_one_records else pd.DataFrame()

# 시계열 라벨링
W_DAYS = 180; SIMILARITY_THRESHOLD_LABEL = 0.6
def parse_news_date(x):
    s = str(x); nums = re.sub(r"[^0-9]", "", s)
    for fmt, cond in [("%Y%m%d%H%M%S", len(nums)==14), ("%Y%m%d%H%M", len(nums)==12), ("%Y%m%d", len(nums)==8)]:
        if cond:
            dt = pd.to_datetime(nums, format=fmt, errors="coerce")
            if pd.notna(dt): return dt
    return pd.to_datetime(s, errors="coerce")

trainable_df_starttime = pd.DataFrame(columns=["topic_id","start_time","end_time","topic_text","label"])
trainable_df_endtime   = pd.DataFrame(columns=["topic_id","start_time","end_time","topic_text","label"])
results_df_label1      = pd.DataFrame(columns=["topic_id","topic_start_date","topic_text","bill_name","bill_date","similarity_score"])

if "date" in df_cleaned.columns and "bill_date" in bills_df.columns and bills_df["bill_date"].notna().any():
    df_topics_only_time = df_cleaned[df_cleaned["topic_id"] != -1].copy()
    df_topics_only_time["date_dt"] = df_topics_only_time["date"].apply(parse_news_date)
    topic_time = (df_topics_only_time.groupby("topic_id", as_index=False).agg(
        start_time=("date_dt","min"), end_time=("date_dt","max"), topic_text=("content", lambda x: " ".join(x.astype(str)))
    ))
    bill_embeddings_timed = embedding_model.encode(bills_df["bill_text"].tolist(), convert_to_tensor=True, show_progress_bar=True)

    def assign_label_with_window(anchor_col, row):
        st = row[anchor_col]
        if pd.isna(st): return 0
        ed = st + pd.Timedelta(days=W_DAYS)
        cands = bills_df[(bills_df["bill_date"] >= st) & (bills_df["bill_date"] <= ed)]
        if cands.empty: return 0
        topic_emb = embedding_model.encode(row["topic_text"], convert_to_tensor=True)
        cand_idx = cands.index.to_list(); cand_emb = bill_embeddings_timed[cand_idx]
        scores = util.pytorch_cos_sim(topic_emb.detach().to("cpu"), cand_emb.detach().to("cpu"))
        return 1 if float(scores.max()) >= SIMILARITY_THRESHOLD_LABEL else 0

    topic_agg_start = topic_time.copy(); topic_agg_start["label"] = topic_agg_start.apply(lambda r: assign_label_with_window("start_time", r), axis=1)
    topic_agg_end   = topic_time.copy(); topic_agg_end["label"]   = topic_agg_end.apply(lambda r: assign_label_with_window("end_time", r), axis=1)

    last_date_in_data = bills_df["bill_date"].max()
    if pd.notna(last_date_in_data):
        cutoff = last_date_in_data - pd.Timedelta(days=W_DAYS)
        trainable_df_starttime = topic_agg_start[topic_agg_start["start_time"] <= cutoff].copy()
        trainable_df_endtime   = topic_agg_end  [topic_agg_end  ["end_time" ] <= cutoff].copy()

        labeled_one = trainable_df_starttime[trainable_df_starttime["label"] == 1]
        if not labeled_one.empty:
            def find_matching_bills_details(row):
                st = row["start_time"]; ed = st + pd.Timedelta(days=W_DAYS)
                cands = bills_df[(bills_df["bill_date"] >= st) & (bills_df["bill_date"] <= ed)]
                if cands.empty: return []
                topic_emb = embedding_model.encode(row["topic_text"], convert_to_tensor=True)
                cand_idx = cands.index.to_list(); cand_emb = bill_embeddings_timed[cand_idx]
                cosine_vec = util.pytorch_cos_sim(topic_emb.detach().to("cpu"), cand_emb.detach().to("cpu"))[0]
                matched = []
                for i in range(len(cands)):
                    sc = float(cosine_vec[i].item())
                    if sc >= SIMILARITY_THRESHOLD_LABEL:
                        b = cands.iloc[i]
                        matched.append({"bill_name": b.get("법률안명", None), "bill_date": b["bill_date"].date(), "similarity_score": sc})
                return sorted(matched, key=lambda x: x["similarity_score"], reverse=True)
            for _, trow in labeled_one.iterrows():
                matches = find_matching_bills_details(trow)
                for m in matches:
                    results_df_label1.loc[len(results_df_label1)] = {
                        "topic_id": int(trow["topic_id"]),
                        "topic_start_date": trow["start_time"].date() if pd.notna(trow["start_time"]) else None,
                        "topic_text": trow.get("topic_text", None),
                        "bill_name": m["bill_name"],
                        "bill_date": m["bill_date"],
                        "similarity_score": m["similarity_score"]
                    }

print("\n[제공되는 주요 DataFrame]")
for name, obj in {
    "df_cleaned": df_cleaned, "df_topics_only": df_topics_only,
    "topic_aggregated_df": topic_aggregated_df, "doc_info_df": doc_info_df,
    "results_df": results_df, "one_to_n_df": one_to_n_df, "n_to_one_df": n_to_one_df,
    "trainable_df_starttime": trainable_df_starttime, "trainable_df_endtime": trainable_df_endtime,
    "results_df_label1": results_df_label1
}.items():
    shape = obj.shape if isinstance(obj, pd.DataFrame) else None
    print(f"- {name}: {'None' if obj is None else shape}")

# =========================
# df_finaltest 생성 (label0 선생성/집계)
# =========================

# 0) 방어
for _n in ["topic_aggregated_df","bills_df","df_cleaned"]:
    if _n not in globals() or globals()[_n] is None or len(globals()[_n]) == 0:
        raise SystemExit(f"[중단] 필수 객체 '{_n}'가 없습니다. 이전 블록 실행 확인 요망.")

# 1) law 라벨
law_label_map = [
    (["정보통신법","정보통신"], "정보통신"),
    (["개인정보보호법","개인정보"], "개인정보"),
    (["아동복지법","아동복지"], "아동복지"),
    (["자본시장법","자본시장"], "자본"),
]
def get_law_label_from_name(name: str) -> str:
    n = (name or "").lower()
    for keys, label in law_label_map:
        if any(k.lower() in n for k in keys):
            return label
    return "기타"

law_value = get_law_label_from_name(file_name)
topic_base = topic_aggregated_df.copy()
if "topic_id" not in topic_base.columns:
    raise ValueError("topic_aggregated_df에 topic_id 컬럼이 없습니다.")
topic_base["law"] = law_value

# 2) trainable_df_starttime 방어
if "trainable_df_starttime" not in globals() or trainable_df_starttime is None:
    trainable_df_starttime = pd.DataFrame(columns=["topic_id","start_time","end_time","topic_text","label"])
if "label" not in trainable_df_starttime.columns:
    trainable_df_starttime["label"] = np.nan
if "topic_id" not in trainable_df_starttime.columns:
    trainable_df_starttime["topic_id"] = pd.Series([], dtype="int64")

# 3) results_df_label1 방어
if "results_df_label1" not in globals() or results_df_label1 is None:
    results_df_label1 = pd.DataFrame(columns=["topic_id","topic_start_date","topic_text","bill_name","bill_date","similarity_score"])

# 4) bills_df 의견수 숫자화
if "의견수" in bills_df.columns:
    def to_num(x):
        try:
            s = str(x).strip().replace(",","")
            if s == "" or s.lower() == "nan":
                return np.nan
            return float(s)
        except:
            return np.nan
    bills_df["_opinion_num"] = bills_df["의견수"].apply(to_num)
else:
    bills_df["_opinion_num"] = np.nan

# 5) lawmake 맵 및 임시 df_tmp
if not trainable_df_starttime.empty and "topic_id" in trainable_df_starttime.columns and "label" in trainable_df_starttime.columns:
    lawmake_map = (trainable_df_starttime[["topic_id","label"]]
                   .dropna(subset=["topic_id"])
                   .astype({"topic_id":"int64"}))
    lawmake_map = lawmake_map.groupby("topic_id", as_index=False)["label"].max()
else:
    lawmake_map = pd.DataFrame(columns=["topic_id","label"])

df_tmp = topic_base.merge(lawmake_map, on="topic_id", how="left").rename(columns={"label":"lawmake"})
df_tmp["lawmake"] = df_tmp["lawmake"].fillna(0).astype(int)

# 6) label 토픽 세트 파생
all_topic_ids = set(df_tmp["topic_id"].astype("int64").tolist())
label1_topic_ids = set(results_df_label1["topic_id"].astype("int64").tolist()) if not results_df_label1.empty else set()
if "label" in trainable_df_starttime.columns:
    label0_topic_ids = set(trainable_df_starttime.loc[trainable_df_starttime["label"]==0, "topic_id"].astype("int64").unique().tolist())
else:
    label0_topic_ids = set(df_tmp.loc[df_tmp["lawmake"]==0, "topic_id"].astype("int64").tolist())
label0_only_topic_ids = sorted((label0_topic_ids - label1_topic_ids))

# 7) results_df_label0 선생성
rows0 = []
for tid in label0_only_topic_ids:
    rows0.append({
        "topic_id": int(tid),
        "topic_start_date": None,
        "topic_text": None,
        "bill_name": None,
        "bill_date": None,
        "similarity_score": 0.0
    })
results_df_label0 = pd.DataFrame(rows0, columns=["topic_id","topic_start_date","topic_text","bill_name","bill_date","similarity_score"])

# 8) 의견수 부착 및 집계
def attach_opinion(df_match: pd.DataFrame) -> pd.DataFrame:
    df = df_match.copy()
    if "bill_name" in df.columns:
        df = df.merge(
            bills_df[["법률안명","_opinion_num"]].rename(columns={"법률안명":"bill_name"}),
            on="bill_name", how="left"
        )
    else:
        df["_opinion_num"] = np.nan
    df["_opinion_num"] = df["_opinion_num"].fillna(0.0)
    return df

r1 = attach_opinion(results_df_label1 if isinstance(results_df_label1, pd.DataFrame) else pd.DataFrame(columns=["topic_id","bill_name","similarity_score"]))
r0 = attach_opinion(results_df_label0)

agg1 = r1.groupby("topic_id", as_index=False).agg(
    opinion_mean_1=("_opinion_num","mean"),
    sim_mean_1=("similarity_score","mean")
) if not r1.empty else pd.DataFrame(columns=["topic_id","opinion_mean_1","sim_mean_1"])

agg0 = r0.groupby("topic_id", as_index=False).agg(
    opinion_mean_0=("_opinion_num","mean"),
    sim_mean_0=("similarity_score","mean")
) if not r0.empty else pd.DataFrame(columns=["topic_id","opinion_mean_0","sim_mean_0"])

# 9) 최종 df_finaltest 생성
df_final = df_tmp.merge(agg1, on="topic_id", how="left")
df_final = df_final.merge(agg0, on="topic_id", how="left")

df_final["opinion"] = np.where(
    df_final["lawmake"]==1, df_final["opinion_mean_1"], df_final["opinion_mean_0"]
)
df_final["similarity_score"] = np.where(
    df_final["lawmake"]==1, df_final["sim_mean_1"], df_final["sim_mean_0"]
)

df_final["opinion"] = df_final["opinion"].fillna(0.0)
df_final["similarity_score"] = df_final["similarity_score"].fillna(0.0)

drop_cols = [c for c in ["opinion_mean_1","opinion_mean_0","sim_mean_1","sim_mean_0"] if c in df_final.columns]
df_finaltest = df_final.drop(columns=drop_cols)

# 10) 진단
print("\n[제공되는 주요 DataFrame]")
for name, obj in {
    "results_df_label1": results_df_label1, "results_df_label0": results_df_label0,
}.items():
    shape = obj.shape if isinstance(obj, pd.DataFrame) else None
    print(f"- {name}: {'None' if obj is None else shape}")

print("\n[df_finaltest] shape:", df_finaltest.shape)
print(df_finaltest.head(5))


# =========================
# 안전장치: B 후속(TF-IDF 기반 소셜→토픽 매칭) 실행 전 점검
# =========================
# 다음 블록이 사용할 전역 객체/컬럼 존재 점검
if "social_df_all" in globals() and social_df_all is not None and len(social_df_all) > 0:
    _HAS_SOCIAL = True
else:
    _HAS_SOCIAL = False
if "topic_aggregated_df" in globals() and topic_aggregated_df is not None and len(topic_aggregated_df) > 0:
    _HAS_TOPIC_AGG = True
else:
    _HAS_TOPIC_AGG = False

print(f"[안전장치] social_df_all OK? -> {_HAS_SOCIAL}")
print(f"[안전장치] topic_aggregated_df OK? -> {_HAS_TOPIC_AGG}")

# =========================
# B-후속: TF-IDF 기반 소셜→토픽 매칭
# =========================
if _HAS_SOCIAL and _HAS_TOPIC_AGG:
    import numpy as np
    from sklearn.feature_extraction.text import TfidfVectorizer
    from sklearn.metrics.pairwise import cosine_similarity

    # ⚙️ 설정
    SOCIAL_PROCESSED_COL = "noun_text"     # social_df_all 컬럼 [web:10]
    TOPIC_ID_COL         = "topic_id"      # topic_aggregated_df 컬럼 [web:10]
    TOPIC_PROCESSED_COL  = "topic_text"    # topic_aggregated_df 컬럼 [web:10]
    SIM_THRESHOLD        = 0.10            # 유사도 임계값 [web:10]

    # 1) 데이터 준비
    social_df = social_df_all.copy()
    topic_df  = topic_aggregated_df.copy()

    if TOPIC_ID_COL in topic_df.columns:
        topic_df = topic_df.dropna(subset=[TOPIC_ID_COL]).copy()
        topic_df[TOPIC_ID_COL] = topic_df[TOPIC_ID_COL].astype(int)
    else:
        raise KeyError(f"토픽 데이터에 '{TOPIC_ID_COL}' 컬럼이 없습니다.")

    for col, name in [(TOPIC_PROCESSED_COL, "토픽 텍스트"), (SOCIAL_PROCESSED_COL, "소셜 텍스트")]:
        if col not in (topic_df.columns if col==TOPIC_PROCESSED_COL else social_df.columns):
            raise KeyError(f"{name} 컬럼 '{col}'을 찾을 수 없습니다.")

    topic_texts  = topic_df[TOPIC_PROCESSED_COL].fillna("")
    social_texts = social_df[SOCIAL_PROCESSED_COL].fillna("")

    print(f"✅ 준비 완료: 토픽 {len(topic_df)}개, 소셜 {len(social_df)}건")

    # 2) TF-IDF 학습 및 변환
    tfidf_vectorizer = TfidfVectorizer(min_df=5, max_df=0.8)  # 너무 희귀/상위 빈도 제거 [web:10]
    combined_texts = pd.concat([topic_texts, social_texts], ignore_index=True)
    tfidf_vectorizer.fit(combined_texts)

    topic_matrix  = tfidf_vectorizer.transform(topic_texts)
    social_matrix = tfidf_vectorizer.transform(social_texts)

    # 3) 코사인 유사도 및 1차 매칭
    cosine_sim = cosine_similarity(social_matrix, topic_matrix)  # (n_social, n_topics)
    matched_topic_indices = np.argmax(cosine_sim, axis=1)

    results_df_tfidf = social_df.copy()
    results_df_tfidf["매칭된_topic_id"] = topic_df.iloc[matched_topic_indices][TOPIC_ID_COL].values
    results_df_tfidf["유사도_점수"] = cosine_sim[np.arange(len(results_df_tfidf)), matched_topic_indices]
    results_df_tfidf = results_df_tfidf.sort_values(by="유사도_점수", ascending=False)

    # 4) 2차 필터링
    threshold = SIM_THRESHOLD
    filtered_df = results_df_tfidf[results_df_tfidf["유사도_점수"] >= threshold].copy()

    print(f"\n--- 📊 2차 필터링 결과 ---")
    print(f"전체 매칭 데이터: {len(results_df_tfidf)}건")
    print(f"유사도 {threshold} 이상 데이터: {len(filtered_df)}건")
    print(f" -> {len(results_df_tfidf) - len(filtered_df)}건 제외")

    # 5) 통계량 계산
    if not filtered_df.empty:
        similarity_stats = filtered_df["유사도_점수"].describe().to_frame()
        print("\n--- 📊 '유사도_점수' 기초 통계량 ---")
        print(similarity_stats)
    else:
        similarity_stats = None
        print("\n--- 📊 필터링된 데이터가 없어 통계량 계산을 건너뜁니다. ---")

    # 메모리에 남기는 객체들
    globals()["results_df_tfidf"] = results_df_tfidf
    globals()["filtered_df_tfidf"] = filtered_df
    globals()["similarity_stats_tfidf"] = similarity_stats
else:
    print("\n[안내] TF-IDF 매칭 블록은 social_df_all 또는 topic_aggregated_df 부재로 건너뛰었습니다.")
# ===========================================
# 통합 파이프라인: 뉴스피처 + 소셜집계 + 최종병합
# (저장 X, 출력만 / 누락 컬럼 자동생성=1 은 중요도 산출 직전에만)
# ===========================================
import os, re, math, subprocess
import numpy as np
import pandas as pd
from collections import Counter
from tqdm import tqdm
import matplotlib.pyplot as plt
from scipy.stats import skew, kurtosis

# -------------------------------------------
# 0) 입력 파일 경로
# -------------------------------------------
NEWS_DF = df_finaltest.copy()
SOCIAL_DF = filtered_df.copy()

# -------------------------------------------
# 1) KNU 감성사전 + 커스텀 사전 결합 (항상 코드 내 커스텀 병합; 파일 I/O 없음)
# -------------------------------------------
if not os.path.exists("/content/KnuSentiLex"):
    try:
        subprocess.run(["git","clone","https://github.com/park1200656/KnuSentiLex.git"], check=True)
    except Exception as e:
        print("⚠️ git clone 실패(이미 있거나 권한 문제일 수 있음):", e)

senti_path = "/content/KnuSentiLex/SentiWord_Dict.txt"
if not os.path.exists(senti_path):
    raise FileNotFoundError("KnuSentiLex 사전 파일을 찾을 수 없습니다.")

senti_df = pd.read_csv(senti_path, sep="\t", names=["word","polarity"], encoding="utf-8").dropna()
senti_df["word"] = senti_df["word"].astype(str).str.strip()

def _safe_float(x):
    try:
        return float(x)
    except Exception:
        try: return float(str(x).strip())
        except Exception: return 0.0

senti_df["polarity"] = senti_df["polarity"].apply(_safe_float)
BASE_LEX = dict(zip(senti_df["word"], senti_df["polarity"]))

# --- 커스텀 사전 (매 실행 병합) ---
POS_LEX = {"좋다":2.0,"최고":2.5,"훌륭":2.0,"행복":2.3,"추천":1.5,"만족":1.8,"즐거움":1.8,"감사":1.7,"괜찮":1.2,
           "멋짐":2.0,"사랑":2.5,"꿀잼":2.4,"디토":1.5,"럭키비키":2.3,"갓생":2.1,"갓성비":2.0,"레전드":2.8,
           "국가권력급":2.6,"킹정":2.1,"인정":1.5,"팩트":1.6,"민주주의":2.2,"투명성":1.8,"개혁":2.0,"혁신":2.1,
           "발전":1.9,"진보":1.7,"통합":2.0,"화합":2.2,"협치":2.1,"소통":1.9,"개방":1.6,"변화":1.8,"공정":2.3,
           "정의":2.4,"평등":2.2,"자유":2.5,"인권":2.3,"복지":2.0,"평화":2.4,"안전":2.1,"안정":1.9,"번영":2.2,
           "성장":1.8,"희망":2.3,"개선":1.9,"향상":1.8,"보완":1.5,"강화":1.7,"확대":1.6,"증진":1.7,"찬성":2.1,
           "지지":2.0,"동의":1.8,"공감":1.9,"응원":2.0,"격려":1.8,"환영":2.2,"축하":2.3,"기대":1.9,"신뢰":2.1,
           "믿음":2.0,"확신":1.9,"합리성":2.0,"타당성":1.9,"필요성":1.7,"중요성":1.8}
NEG_LEX = {"나쁨":-2.0,"최악":-2.5,"실망":-2.0,"짜증":-2.0,"불만":-1.8,"혐오":-2.3,"반감":-2.0,"불편":-1.6,
           "분노":-2.0,"킹받음":-2.2,"노잼":-2.1,"현타":-1.8,"멘붕":-1.9,"억텐":-1.5,"독재":-2.8,"전제":-2.6,
           "독단":-2.4,"폭정":-2.7,"억압":-2.5,"탄압":-2.6,"부패":-2.7,"비리":-2.5,"부정":-2.3,"편법":-2.2,
           "불법":-2.6,"위법":-2.4,"사기":-2.5,"거짓":-2.3,"기만":-2.4,"조작":-2.6,"왜곡":-2.2,"은폐":-2.4,
           "무능":-2.3,"무책임":-2.4,"무관심":-1.8,"무시":-2.1,"방관":-1.9,"외면":-1.8,"갈등":-2.2,"분열":-2.4,
           "대립":-2.1,"충돌":-2.3,"마찰":-1.9,"반목":-2.0,"위기":-2.4,"혼란":-2.2,"불안":-2.1,"우려":-1.8,
           "걱정":-1.7,"염려":-1.6,"반대":-2.1,"거부":-2.2,"거절":-2.0,"비판":-1.8,"비난":-2.3,"성토":-2.4,
           "규탄":-2.5,"의혹":-2.0,"의심":-1.9,"불신":-2.2,"배신":-2.4,"실패":-2.2,"좌절":-2.1,"포기":-1.9,
           "철회":-1.8,"취소":-1.7,"중단":-1.6,"극좌":-2.4,"극우":-2.4,"종북":-2.6,"토착왜구":-2.7,"친일":-2.5,
           "매국":-2.8,"빨갱이":-2.7,"좌빨":-2.6,"수구":-2.3,"꼴통":-2.4,"수꼴":-2.5,"일베":-2.2,"개독":-2.5,
           "한남":-2.3,"페미":-2.1,"웜페":-2.4,"메갈":-2.3,"래디컬":-2.2,"극혐":-2.8,"개노답":-2.3,"쓰레기":-2.4,
           "헬조선":-2.3,"계엄":-2.7,"쿠데타":-2.9,"내란":-2.8,"반란":-2.7,"반역":-2.8,"스캔들":-2.2,"게이트":-2.1}

def combine_lexicons(lib_dict, custom_dict, mode="priority_custom"):
    out = lib_dict.copy()
    for w, sc in custom_dict.items():
        if w in out:
            out[w] = sc if mode!="max_abs" else (sc if abs(sc) >= abs(out[w]) else out[w])
        else:
            out[w] = sc
    return out

LEXICON = combine_lexicons(BASE_LEX, {**POS_LEX, **NEG_LEX}, mode="priority_custom")
print(f"🧩 커스텀 사전 {len(POS_LEX)+len(NEG_LEX):,}개 병합 완료 (최종 어휘 수: {len(LEXICON):,})")

def sentiment_score_tokens(tokens):
    if not tokens: return 0.0
    return float(sum(LEXICON.get(tok, 0.0) for tok in tokens))

def sentiment_score_ws(text: str) -> float:
    if not isinstance(text, str): return 0.0
    return float(sum(LEXICON.get(tok, 0.0) for tok in text.split()))

# -------------------------------------------
# 2) 토크나이저
# -------------------------------------------
try:
    from konlpy.tag import Okt
    okt = Okt()
    def tokenize(text):
        if not isinstance(text, str) or not text.strip(): return []
        return [t for t in okt.morphs(text, stem=True) if t.strip()]
except Exception:
    def tokenize(text):
        if not isinstance(text, str) or not text.strip(): return []
        return re.findall(r"[가-힣]+|[A-Za-z]+|\d+", text)

# -------------------------------------------
# 3) 뉴스 피처
# -------------------------------------------
def sentiment_features(tokens):
    if not tokens: return 0.0, 0, 0, 0.0
    scores = [LEXICON.get(t, 0.0) for t in tokens]
    score_sum = float(sum(scores))
    pos_cnt = int(sum(1 for s in scores if s > 0))
    neg_cnt = int(sum(1 for s in scores if s < 0))
    ratio = pos_cnt / (pos_cnt + neg_cnt + 1e-6)
    return score_sum, pos_cnt, neg_cnt, ratio

def repetition_features(tokens, min_freq=2, topn=3):
    if not tokens: return 0, 0, "", 0, "", 0.0
    c = Counter(tokens)
    rep_items = [(w, f) for w, f in c.items() if f >= min_freq]
    rep_kw_count = len(rep_items)
    rep_kw_total = sum(max(f - 1, 0) for _, f in rep_items)
    top_items = sorted(c.items(), key=lambda x: (-x[1], x[0]))[:topn]
    rep_kw_top1 = top_items[0][0] if top_items else ""
    rep_kw_top1_freq = top_items[0][1] if top_items else 0
    rep_kw_top3 = ",".join([w for w, _ in top_items]) if top_items else ""
    n = sum(c.values())
    probs = [f / n for f in c.values()]
    entropy = -sum(p * math.log(p + 1e-12) for p in probs) if n > 0 else 0.0
    return rep_kw_count, rep_kw_total, rep_kw_top1, rep_kw_top1_freq, rep_kw_top3, entropy

def simple_text_features(text, tokens):
    if not isinstance(text, str): text = ""
    char_len = len(text); word_len = len(tokens)
    uniq_words = len(set(tokens)) if tokens else 0
    ttr = uniq_words / (word_len + 1e-6)
    avg_tok_len = np.mean([len(t) for t in tokens]) if tokens else 0
    num_cnt = sum(1 for t in tokens if t.isdigit())
    eng_cnt = sum(1 for t in tokens if re.fullmatch(r"[A-Za-z]+", t))
    num_ratio = num_cnt / (word_len + 1e-6)
    eng_ratio = eng_cnt / (word_len + 1e-6)
    q_count = text.count("?"); exc_count = text.count("!")
    c = Counter(tokens)
    rep_kw_density = sum(1 for f in c.values() if f >= 2) / (word_len + 1e-6)
    return {"char_len":char_len,"word_len":word_len,"uniq_words":uniq_words,"ttr":ttr,
            "avg_tok_len":avg_tok_len,"num_ratio":num_ratio,"eng_ratio":eng_ratio,
            "punc_q":q_count,"punc_exc":exc_count,"rep_kw_density":rep_kw_density}

def build_news_features(df_raw: pd.DataFrame) -> pd.DataFrame:
    df = df_raw.copy()
    if "topic_text" not in df.columns:  # 자동 생성하지 않음
        cands = [c for c in df.columns if c.lower() in {"topic_text","text","content","full_text","본문"}]
        if not cands:
            raise KeyError(f"'topic_text' 컬럼을 찾을 수 없음: {list(df.columns)}")
        df = df.rename(columns={cands[0]: "topic_text"})
    if "topic_id" not in df.columns:
        df["topic_id"] = np.arange(len(df))
    df["tokens"] = df["topic_text"].apply(tokenize)

    sent = df["tokens"].apply(sentiment_features)
    sent_df = pd.DataFrame(sent.tolist(), columns=["sent_score","sent_pos","sent_neg","sent_ratio"])
    df = pd.concat([df, sent_df], axis=1)

    rep = df["tokens"].apply(lambda t: repetition_features(t, min_freq=2, topn=3))
    df["rep_kw_count"]     = rep.apply(lambda x: x[0])
    df["rep_kw_total"]     = rep.apply(lambda x: x[1])
    df["rep_kw_top1"]      = rep.apply(lambda x: x[2])
    df["rep_kw_top1_freq"] = rep.apply(lambda x: x[3])
    df["rep_kw_top3"]      = rep.apply(lambda x: x[4])
    df["entropy"]          = rep.apply(lambda x: x[5])

    basic = df.apply(lambda r: simple_text_features(r["topic_text"], r["tokens"]),
                     axis=1, result_type="expand")
    df = pd.concat([df, basic], axis=1)
    df["sent_per_100w"] = df.apply(lambda r: 100 * r["sent_score"] / (r["word_len"] + 1e-6), axis=1)
    return df.drop(columns=["tokens"])

# -------------------------------------------
# 4) law / law_group 라벨링
# -------------------------------------------
def law_from_path(path: str) -> str:
    name = os.path.basename(path)
    if "아동" in name: return "아동복지"
    if "자본" in name: return "자본"
    if "개인정보" in name: return "개인정보"
    if "정통법" in name or "정보통신" in name: return "정보통신"
    return "기타"

def law_to_group(law: str) -> str:
    return "개인정보_정보통신" if law in {"개인정보","정보통신"} else law

def social_group_from_path(path: str) -> str:
    name = os.path.basename(path)
    if "개인정보" in name and "정보통신" in name: return "개인정보_정보통신"
    if "아동복지" in name: return "아동복지"
    if "자본" in name: return "자본"
    return "기타"

# -------------------------------------------
# 5) 소셜 표준화/집계 (필수열 미존재 시 자동 생성하지 않음)
# -------------------------------------------
CANDIDATES = {
    "매칭된_topic_id": ["매칭된_topic_id","matched_topic_id","topic_id"],
    "date":            ["date","작성일시","created_at","datetime"],
    "source":          ["source","platform","origin"],
    "본문_clean":      ["본문_clean","본문","content_clean","text_clean","full_text","fulltext"],
    "noun_text":       ["noun_text","noun_tokens","명사열","명사_토큰"],
    "유사도_점수":     ["유사도_점수","similarity_score","sim_score"],
}

def normalize_columns(df: pd.DataFrame) -> pd.DataFrame:
    df = df.copy()
    rename_map = {}
    for std, cands in CANDIDATES.items():
        for c in cands:
            if c in df.columns:
                rename_map[c] = std; break
    df = df.rename(columns=rename_map)
    req = ["매칭된_topic_id","date","source","본문_clean","noun_text"]
    miss = [c for c in req if c not in df.columns]
    if miss:  # 자동 생성하지 않음
        raise KeyError(f"필수 컬럼 누락: {miss} / 실제: {list(df.columns)}")
    return df

def parse_datetime(series: pd.Series) -> pd.Series:
    s = series.astype(str)
    dt = pd.to_datetime(s, format="%Y.%m.%d.%H", errors="coerce")
    if dt.isna().all(): dt = pd.to_datetime(s, errors="coerce")
    return dt

def process_social_data(df: pd.DataFrame) -> pd.DataFrame:
    df = normalize_columns(df)
    df["datetime"]  = parse_datetime(df["date"])
    df = df.dropna(subset=["datetime"]).copy()
    df["date_only"] = df["datetime"].dt.date
    df["hour"]      = df["datetime"].dt.hour
    df["weekday"]   = df["datetime"].dt.weekday
    df["is_weekend"]= df["weekday"] >= 5
    df["text_len"]  = df["본문_clean"].astype(str).apply(len)
    df["sentiment"] = df["noun_text"].astype(str).apply(sentiment_score_ws)

    rows = []
    for topic, g in tqdm(df.groupby("매칭된_topic_id"), desc="소셜 토픽 집계"):
        daily = g.groupby("date_only").size().rename("count").to_frame()
        daily.index = pd.to_datetime(daily.index)
        daily = daily.asfreq("D", fill_value=0)
        mu, sigma = daily["count"].mean(), daily["count"].std()
        daily["z_post"] = (daily["count"] - mu) / (sigma + 1e-6)
        daily["ewma_post"] = daily["count"].ewm(span=3, adjust=False).mean()
        ewma_slope = daily["ewma_post"].diff().mean()

        peak_date = daily["count"].idxmax() if not daily.empty else pd.NaT
        first_date = daily.index.min() if not daily.empty else pd.NaT
        days_to_peak = int((peak_date - first_date).days) if pd.notna(peak_date) and pd.notna(first_date) else np.nan
        period_days = int((daily.index.max() - daily.index.min()).days) + 1 if not daily.empty else 0

        rows.append({
            "topic_id":        int(topic),
            "period_days":     period_days,
            "total_posts":     int(len(g)),
            "unique_sources":  int(g["source"].nunique()),
            "source_diversity": g["source"].nunique() / (len(g) + 1e-6),
            "avg_len":         float(g["text_len"].mean()),
            "weekend_ratio":   float(g["is_weekend"].mean()),
            "mode_hour":       int(g["hour"].mode()[0]) if not g["hour"].mode().empty else np.nan,
            "avg_sentiment":   float(g["sentiment"].mean()),
            "max_zscore":      float(daily["z_post"].max()) if not daily.empty else np.nan,
            "ewma_slope":      float(ewma_slope) if not daily.empty else np.nan,
            "days_to_peak":    days_to_peak,
            "peak_date":       peak_date.date() if pd.notna(peak_date) else np.nan,
            "mean_similarity": float(g["유사도_점수"].mean()) if "유사도_점수" in g.columns else np.nan,
            "max_similarity":  float(g["유사도_점수"].max())  if "유사도_점수" in g.columns else np.nan,
        })
    return pd.DataFrame(rows)

# 6) 뉴스/소셜 처리 + 병합  (CSV X, 메모리 DF 사용)
news_all = build_news_features(NEWS_DF)

# law / law_group 지정
# - df_finaltest에 'law' 컬럼이 있으면 그대로 사용
# - 없으면 기본값 '기타' (원하면 '자본' 등으로 바꿔도 됩니다)
if "law" in NEWS_DF.columns and not NEWS_DF["law"].dropna().empty:
    _news_law = str(NEWS_DF["law"].iloc[0])
else:
    _news_law = "기타"   # <- 필요 시 "자본" 등으로 변경

news_all["law"] = _news_law
news_all["law_group"] = law_to_group(_news_law)

# 소셜 처리
social_all = process_social_data(SOCIAL_DF)

# 소셜 DF에 'law_group'이 이미 있으면 그걸 사용, 없으면 뉴스의 law_group을 따라가게
if "law_group" in SOCIAL_DF.columns and not SOCIAL_DF["law_group"].dropna().empty:
    _social_group = str(SOCIAL_DF["law_group"].iloc[0])
else:
    _social_group = news_all["law_group"].iloc[0]

social_all["law_group"] = _social_group

# 병합
merge_keys = ["topic_id","law_group"]
df = pd.merge(
    news_all, social_all, how="left", on=merge_keys, suffixes=("_news","_social")
)


# -------------------------------------------
# All-in-One: 중요도 산출(0~100/0~1)  — 여기서만 누락 컬럼 자동 생성=1
# (파라미터 고정 + 사용 컬럼 리포트 포함)
# -------------------------------------------

# ===== 1) 고정 파라미터 =====
AXIS_W = {  # 축 가중치
    "Buzz": 0.30,
    "Focus": 0.25,
    "Depth": 0.25,
    "Spread": 0.20,
    "Penalty": 0.15,  # 최종에서 빼는 항목
}
INT_W = {   # 내부 가중(축 내부 조합 시)
    "puncq": 0.5,     # Buzz 내부의 punc_q 가중
    "uniq":  0.3,     # Focus 내부의 uniq_words 가중
}
# 모델이 기대하는 '기본 피처명' (pick 규칙: *_social > *_news > base)
FEATURES = {
    "posts":  "total_posts",      # Buzz
    "zpeak":  "max_zscore",       # Buzz
    "puncq":  "punc_q",           # Buzz
    "entropy":"entropy",          # Focus
    "repct":  "rep_kw_count",     # Focus
    "reptt":  "rep_kw_total",     # Focus
    "uniq":   "uniq_words",       # Focus
    "atok":   "avg_tok_len",      # Depth
    "alen":   "avg_len",          # Depth
    "sr":     "sent_ratio",       # Depth + Senti
    "sp100":  "sent_per_100w",    # Depth + Senti
    "ssc":    "sent_score",       # Senti
    "slope":  "ewma_slope",       # Spread
    "dpeak":  "days_to_peak",     # Penalty + speed 계산
    # Spread의 source_entropy_norm 계산용 대체 입력(둘 중 1개만 있어도 OK)
    "src_diversity": "source_diversity",
    "unique_sources":"unique_sources",
}

def zscore(s):
    s = pd.to_numeric(s, errors="coerce")
    med = s.median(skipna=True); s = s.fillna(med)
    mu = s.mean(); sd = s.std()
    return (s - mu) / (sd + 1e-9)

def pick(df, base):
    for n in (f"{base}_social", f"{base}_news", base):
        if n in df.columns: return n
    return None  # 못 찾으면 None

def ensure_columns(df: pd.DataFrame, bases: list):
    """중요도 산출에 필요한 '기본 피처'가 없으면 여기서만 생성(=1)"""
    for base in bases:
        # pick으로 못 찾을 때만 생성
        if pick(df, base) is None:
            print(f"@@열이 존재하지 않아 생성됨: {base}")
            df[base] = 1
    return df

# 2) 누락 컬럼 보장(=1) — '기본 피처'만 대상(파생값은 아래에서 계산)
needed_raw = [
    FEATURES["posts"], FEATURES["zpeak"], FEATURES["puncq"],
    FEATURES["entropy"], FEATURES["repct"], FEATURES["reptt"], FEATURES["uniq"],
    FEATURES["atok"], FEATURES["alen"], FEATURES["sr"], FEATURES["sp100"], FEATURES["ssc"],
    FEATURES["slope"], FEATURES["dpeak"],
    FEATURES["src_diversity"], FEATURES["unique_sources"],
]
df = ensure_columns(df, needed_raw)

# 3) 타입 캐스팅(안전)
for c in set(needed_raw):
    if c in df.columns:
        df[c] = pd.to_numeric(df[c], errors="coerce")

# 4) 'Resolved columns' 선택(실제 쓰일 최종 컬럼명)
resolved = {k: pick(df, base) for k, base in FEATURES.items()}

# 5) 파생값 계산: source_entropy_norm, speed_intensity
#    - srcH = 1 - z(diversity-like)
_div_col = resolved.get("src_diversity") or resolved.get("unique_sources")
if _div_col:
    z_div = zscore(df[_div_col])
    srcH = 1 - z_div
else:
    # 둘 다 없으면 상수열(=1) → z=0 → srcH=1
    srcH = pd.Series(1 - zscore(pd.Series([1]*len(df))), index=df.index)

#    - speed = posts / (dpeak+1)
_posts_col = resolved.get("posts")
_dpeak_col = resolved.get("dpeak")
if _posts_col and _dpeak_col:
    speed = pd.to_numeric(df[_posts_col], errors="coerce") / (pd.to_numeric(df[_dpeak_col], errors="coerce") + 1)
else:
    speed = pd.Series(np.nan, index=df.index)

# 6) 감성 강도: mean(z(ssc), z(sr), z(sp100))
def z_of(colname, fallback_val=1):
    if not colname:
        return zscore(pd.Series([fallback_val]*len(df)))
    return zscore(df[colname])

z_ssc   = z_of(resolved.get("ssc"))
z_sr    = z_of(resolved.get("sr"))
z_sp100 = z_of(resolved.get("sp100"))
sent_intensity = (z_ssc + z_sr + z_sp100) / 3.0

# 7) 축 점수 계산(정의 고정)
# Buzz  = ( z(posts) + z(zpeak) + sent_intensity + 0.5*z(puncq) ) / 3.5
z_posts = z_of(resolved.get("posts"))
z_zpeak = z_of(resolved.get("zpeak"))
z_puncq = z_of(resolved.get("puncq"))
Buzz = (z_posts + z_zpeak + sent_intensity + INT_W["puncq"] * z_puncq) / (1 + 1 + 1 + INT_W["puncq"])

# Focus = ( 0.5*z(entropy) + 0.5*mean(z(repct), z(reptt)) + 0.3*z(uniq) ) / 1.3
z_entropy = z_of(resolved.get("entropy"))
z_repct   = z_of(resolved.get("repct"))
z_reptt   = z_of(resolved.get("reptt"))
z_uniq    = z_of(resolved.get("uniq"))
Focus = (0.5 * z_entropy + 0.5 * (z_repct + z_reptt) / 2.0 + INT_W["uniq"] * z_uniq) / (0.5 + 0.5 + INT_W["uniq"])

# Depth = mean( z(atok), z(alen), z(sr), z(sp100) )
z_atok = z_of(resolved.get("atok"))
z_alen = z_of(resolved.get("alen"))
Depth  = (z_atok + z_alen + z_sr + z_sp100) / 4.0

# Spread= mean( z(srcH), z(max(0,slope)), z(speed) )
_slope_col = resolved.get("slope")
slope = pd.to_numeric(df[_slope_col], errors="coerce") if _slope_col else pd.Series(np.nan, index=df.index)
slope = slope.clip(lower=0)  # 음수 → 0
z_srcH  = zscore(srcH)
z_slope = zscore(slope)
z_speed = zscore(speed)
Spread  = (z_srcH + z_slope + z_speed) / 3.0

# Penalty = z( 1/(dpeak+1) )
if _dpeak_col:
    days = pd.to_numeric(df[_dpeak_col], errors="coerce")
    Penalty = zscore(1 / (days + 1))
else:
    Penalty = pd.Series(0, index=df.index, dtype=float)

# 8) 최종 점수(가중합) + 0~1/0~100 변환
raw = AXIS_W["Buzz"]*Buzz + AXIS_W["Focus"]*Focus + AXIS_W["Depth"]*Depth + AXIS_W["Spread"]*Spread - AXIS_W["Penalty"]*Penalty
mn, mx = np.nanmin(raw), np.nanmax(raw)
score_01 = (raw - mn) / (mx - mn) if (np.isfinite(mn) and np.isfinite(mx) and mx > mn) else (raw - mn)
df["importance_score"] = score_01.clip(0, 1)
df["law_demand_index"] = (df["importance_score"] * 100).round(2)

# 9) 사용 컬럼/가중치 리포트(저장 X, 출력만)
def _resolved_or_placeholder(k):
    col = resolved.get(k)
    return col if col is not None else f"(auto-filled 1) <- {FEATURES[k]}"

print("\n--- [Importance Scoring Manifest] ---")
print("Axis Weights:", AXIS_W)
print("Internal Weights:", INT_W)
print("\nResolved Columns:")
for k in ["posts","zpeak","puncq","entropy","repct","reptt","uniq","atok","alen","sr","sp100","ssc","slope","dpeak","src_diversity","unique_sources"]:
    print(f"  {k:>12}: {_resolved_or_placeholder(k)}")

# 10) 라벨(영문) 및 출력용 뷰
law_map_en = {"개인정보":"Personal Data","정보통신":"Telecom","아동복지":"Child Welfare","자본":"Capital Market"}
df["law_en"] = df["law"].map(law_map_en).fillna(df["law"].astype(str))
viz = df.copy()

# ===== 출력만 =====
pd.set_option("display.max_columns", 200); pd.set_option("display.width", 200)

print("\n=== [Merged DF Head] ===")
print(df.head(10))

print("\n=== [Importance Preview] ===")
print(df[["topic_id","law","importance_score","law_demand_index"]].head(20))

try:
    plt.figure(figsize=(6,4))
    plt.hist(df["importance_score"].dropna(), bins=20, alpha=0.8,
            color='gold', edgecolor='black')
    plt.title("Importance Score Distribution"); plt.xlabel("Importance (0–1)"); plt.ylabel("Count")
    plt.tight_layout(); plt.show()
except Exception as e:
    print("히스토그램 표시 중 오류:", e)

# ===== 기술통계 출력 =====
stats_df = viz.copy()
imp = stats_df["importance_score"].dropna()
overall = imp.describe(percentiles=[0.1,0.25,0.5,0.75,0.9]).to_frame(name="importance_score")
overall.loc["skew"] = skew(imp, nan_policy="omit")
overall.loc["kurtosis"] = kurtosis(imp, nan_policy="omit", fisher=True)

def _gini_series(s):
    s = s.dropna().values.astype(float)
    if s.size == 0: return np.nan
    s_sorted = np.sort(s); cum = np.cumsum(s_sorted)
    n = len(s_sorted)
    return (n + 1 - 2*np.sum(cum)/cum[-1]) / n if cum[-1] != 0 else 0.0

overall.loc["gini"] = _gini_series(stats_df["importance_score"])
for thr in [0.8,0.9,0.95]:
    overall.loc[f"share_ge_{thr}"] = (imp >= thr).mean()

print("\n=== [Overall Stats for Importance] ===")
print(overall)

rows = []
for law_g, sub in stats_df.groupby("law_en", dropna=True):
    arr = sub["importance_score"].dropna().values
    if len(arr) == 0: continue
    s_sorted = np.sort(arr); cum = np.cumsum(s_sorted); n = len(arr)
    rows.append({
        "law": law_g,
        "count": len(arr),
        "mean": float(np.mean(arr)),
        "std": float(np.std(arr, ddof=1)) if len(arr) > 1 else np.nan,
        "min": float(np.min(arr)),
        "p10": float(np.percentile(arr,10)),
        "p25": float(np.percentile(arr,25)),
        "median": float(np.median(arr)),
        "p75": float(np.percentile(arr,75)),
        "p90": float(np.percentile(arr,90)),
        "max": float(np.max(arr)),
        "skew": float(skew(arr)) if len(arr)>2 else np.nan,
        "kurtosis": float(kurtosis(arr, fisher=True)) if len(arr)>3 else np.nan,
        "gini": ( (n + 1 - 2*np.sum(cum)/cum[-1]) / n ) if cum[-1] != 0 else 0.0,
        "share_ge_0.8": float((arr >= 0.8).mean()),
        "share_ge_0.9": float((arr >= 0.9).mean()),
        "share_ge_0.95": float((arr >= 0.95).mean()),
    })
bylaw_df = pd.DataFrame(rows).sort_values("mean", ascending=False) if rows else pd.DataFrame()
print("\n=== [By-Law Summary for Importance] ===")
print(bylaw_df if not bylaw_df.empty else "(no data)")
```
